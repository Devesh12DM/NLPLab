```python
!pip install nltk spacy networkx matplotlib
!python -m spacy download en_core_web_sm

import nltk
import spacy
import networkx as nx
import matplotlib.pyplot as plt

from nltk.tokenize import word_tokenize
from nltk import pos_tag, RegexpParser

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

nlp = spacy.load("en_core_web_sm")

sentences = [
    "The little girl sings a beautiful song.",
    "The boy is playing cricket.",
    "Students are learning natural language processing.",
    "The cat sleeps on the mat."
]

def rule_based_pos(sentence):
    tokens = word_tokenize(sentence)
    tagged = pos_tag(tokens)

    grammar = r"""
        NP: {<DT>?<JJ.*>*<NN.*>+}
        VP: {<VB.*>+}
        ADJP: {<JJ.*>+}
        ADVP: {<RB.*>+}
    """

    parser = RegexpParser(grammar)
    return parser.parse(tagged)

def statistical_pos(sentence):
    tokens = word_tokenize(sentence)
    return pos_tag(tokens)

def dependency_parsing(sentence):
    doc = nlp(sentence)
    G = nx.DiGraph()

    for token in doc:
        G.add_node(token.text)

        if token.head != token:
            G.add_edge(token.head.text, token.text,
                       label=token.dep_)

    plt.figure(figsize=(10, 5))
    pos = nx.spring_layout(G, seed=42)

    nx.draw(
        G, pos,
        with_labels=True,
        node_color="lightblue",
        node_size=2500,
        font_size=10,
        arrows=True
    )

    edge_labels = nx.get_edge_attributes(G, 'label')
    nx.draw_networkx_edge_labels(
        G, pos,
        edge_labels=edge_labels,
        font_color="red"
    )

    plt.title("Dependency Graph")
    plt.show()

    for token in doc:
        print(token.text, "-->", token.dep_,
              "-->", token.head.text)

def phrase_structure(sentence):
    grammar = r"""
        NP: {<DT>?<JJ.*>*<NN.*>+}
        PP: {<IN><NP>}
        VP: {<VB.*><NP|PP>*}
    """

    parser = RegexpParser(grammar)
    tagged = pos_tag(word_tokenize(sentence))
    tree = parser.parse(tagged)

    print(tree)
    tree.pretty_print()

for i, sentence in enumerate(sentences, 1):

    print("\n", "="*60)
    print("SENTENCE", i, ":", sentence)
    print("="*60)

    print("\n1. RULE-BASED POS TAGGING")
    print(rule_based_pos(sentence))

    print("\n2. STATISTICAL POS TAGGING")
    for word, tag in statistical_pos(sentence):
        print(word, ":", tag)

    print("\n3. DEPENDENCY PARSING")
    dependency_parsing(sentence)

    print("\n4. PHRASE STRUCTURE TREE")
    phrase_structure(sentence)

print("\nALL NLP ANALYSIS COMPLETED SUCCESSFULLY!")
```
