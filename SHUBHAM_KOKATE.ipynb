```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
text = "Steve Jobs founded Apple in California. Sundar Pichai works at Google."

doc = nlp(text)
```

```python
print("Named Entities:")

for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
print("Dependency Parsing:")

for token in doc:
    print(token.text, token.dep_, token.head.text)
```

```python
print("Extracted Relationships:")

entities = [ent for ent in doc.ents]

for ent1 in entities:
    for ent2 in entities:
        if ent1 != ent2 and ent1.start < ent2.start:
            path = []

            for token in doc[ent1.start:ent2.end]:
                path.append(token.text)

            relation = " ".join(path)

            print(ent1.text, "-->", relation, "-->", ent2.text)
```

```python
def subsequence_kernel(path1, path2):
    common = 0

    for word in path1:
        if word in path2:
            common += 1

    return common

path1 = ["Steve", "Jobs", "founded", "Apple"]
path2 = ["Sundar", "Pichai", "works", "Google"]

print("Subsequence Kernel Score:",
      subsequence_kernel(path1, path2))
```

```python
print("Relation Extraction Result:")

for ent1 in entities:
    for ent2 in entities:
        if ent1 != ent2 and ent1.start < ent2.start:
            relation = [
                token.text for token in doc[ent1.start:ent2.end]
            ]

            print(ent1.text, "->", " ".join(relation), "->", ent2.text)
```
