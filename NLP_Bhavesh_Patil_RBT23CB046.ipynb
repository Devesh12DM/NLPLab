```python
#RBT23CB046


import spacy
from spacy import displacy

# Load SpaCy model
try:
    nlp = spacy.load('en_core_web_sm')
except OSError:
    print('Downloading SpaCy model "en_core_web_sm"...')
    !python -m spacy download en_core_web_sm
    nlp = spacy.load('en_core_web_sm')

print('SpaCy setup complete.')
```

```python
sentence = "t"
print(f"Original sentence: {sentence}")
```

```python
print("\n--- SpaCy POS Tagging ---")
doc_spacy = nlp(sentence)
pos_tags_spacy = [(token.text, token.pos_) for token in doc_spacy]
print(pos_tags_spacy)
```

```python
print("\n--- SpaCy Dependency Parsing ---")
for token in doc_spacy:
    print(f"{token.text:{10}} {token.dep_:{10}} {token.head.text:{10}} {token.pos_:{5}} {[child.text for child in token.children]}")

print("\n--- Visualizing Dependency Graph ---")
displacy.render(doc_spacy, style="dep", jupyter=True, options={'compact': True, 'distance': 90})




























```
